```python
import os
import requests
import json
from datetime import datetime
```

```python
model = "gpt-3.5-turbo"
prompt_type = "simple_prompt"
```

```python
# make directory with model name in current directory and then create a prompt_type directory inside it
os.makedirs(os.path.join(os.getcwd(), model, prompt_type), exist_ok=True)

    
```

```python
# create a log file in the prompt_type directory with name "tmp_1.log"
log_file_path = os.path.join(os.getcwd(), model, prompt_type, "tmp_1.log")
with open(log_file_path, "w") as log_file:
    log_file.write("Log file created at: " + str(datetime.now()) + "\n")
    log_file.write("Model: " + model + "\n")
    log_file.write("Prompt Type: " + prompt_type + "\n")
    log_file.write("Prompt: " + prompt_type + "\n")
    log_file.write("Response: \n\n")
```

```python
log_data = {
    "timestamp": datetime.now().isoformat(),
    "model": model,
    "prompt": prompt_type,
    "output": prompt_type
}
log_file = "llm_logs.json"
```

```python
log_path = os.path.join(os.getcwd(), model, prompt_type, log_file)

# Ensure the directory exists
os.makedirs(os.path.dirname(log_path), exist_ok=True)

# If the file doesn't exist, create it with initial log_data
if not os.path.exists(log_path):
    with open(log_path, "w") as f:
        json.dump([log_data], f, indent=4)
else:
    with open(log_path, "r+") as f:
        try:
            data = json.load(f)
        except json.JSONDecodeError:
            data = []

        data.append(log_data)
        f.seek(0)
        json.dump(data, f, indent=4)
        f.truncate()
```

```python
with open(os.path.join(os.getcwd(), model, prompt_type, log_file), "w") as json_file:
    json.dump(log_data, json_file, indent=4)
    
with open(os.path.join(os.getcwd(), model, prompt_type, log_file), "a") as f:
    f.write(json.dumps(log_data) + "\n")
    
if not os.path.exists(log_file):
    with open(log_file, "w") as f:
        json.dump([log_data], f, indent=4)
else:
    with open(log_file, "r+") as f:
        data = json.load(f)
        data.append(log_data)
        f.seek(0)
        json.dump(data, f, indent=4)
```
